# Credit events

Attach a `CreditEventEmitter` when constructing `Bursar`. Capability operations emit application events after their canonical ledger mutation succeeds.

```python
from bursar import Bursar, CreditEventEmitter
from bursar.stores.memory import MemoryStore

emitter = CreditEventEmitter()
bursar = Bursar.create(credit_store=MemoryStore(), emitter=emitter)
print(type(bursar.credits).__name__)
```
