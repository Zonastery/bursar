# Credit expiry

Positive ledger entries create lots. Lot expiry and allocation are the sole source of expiring bucket availability; expiry posts a canonical ledger entry rather than updating a projection.

```python
from datetime import UTC, datetime, timedelta
import uuid
from bursar import Bursar
from shared import start_postgres_store, cleanup

store, pgdata = start_postgres_store()
try:
    bursar = Bursar.create(credit_store=store)
    user_id = str(uuid.uuid4())
    result = bursar.credits.add_credits(user_id, 25, expires_at=datetime.now(UTC) + timedelta(days=30))
    print(result.entry_id)
finally:
    cleanup(pgdata)
```
